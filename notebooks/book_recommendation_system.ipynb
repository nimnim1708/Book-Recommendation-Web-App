# Top 50 popular Books

```python
# Combine ratings with book metadata (title/author/image)
ratings_with_name = ratings.merge(books,on='ISBN')
```

```python
# Count how many ratings each book received
num_rating_df = ratings_with_name.groupby('Book-Title').count()['Book-Rating'].reset_index()
num_rating_df.rename(columns={'Book-Rating':'num_ratings'},inplace=True)
num_rating_df
```

```python
# Compute average rating per book
avg_rating_df = ratings_with_name.groupby('Book-Title')['Book-Rating'].mean().reset_index()
avg_rating_df.rename(columns={'Book-Rating':'avg_rating'},inplace=True)
avg_rating_df
```

```python
# Merge count + average into one table
popular_df = num_rating_df.merge(avg_rating_df,on='Book-Title')
popular_df
```

```python
# Keep only books with many ratings and take top 50
popular_df = popular_df[popular_df['num_ratings']>=100].sort_values('avg_rating',ascending=False).head(50)
popular_df
```

```python
popular_df = popular_df.merge(books,on='Book-Title').drop_duplicates('Book-Title')[['Book-Title','Book-Author','Image-URL-L','num_ratings','avg_rating']]
popular_df
```

# Collaborative filtering based recommender system

```python
# Keep only "active users" (users who rated many books)
x = ratings_with_name.groupby('User-ID').count()['Book-Rating'] > 20
active_users = x[x].index
```

```python
filtered_rating = ratings_with_name[ratings_with_name['User-ID'].isin(active_users)]
```

```python
# Keep only "popular books" (books with enough ratings)
y = filtered_rating.groupby('Book-Title').count()['Book-Rating']>=10
famous_books = y[y].index
final_ratings = filtered_rating[filtered_rating['Book-Title'].isin(famous_books)]
```

```python
# Pivot table: rows = books, columns = users, values = ratings
pt = final_ratings.pivot_table(index='Book-Title',columns='User-ID',values='Book-Rating')
pt.fillna(0,inplace=True)
pt
```

```python
# Compute cosine similarity between all book vectors (706x706)
from sklearn.metrics.pairwise import cosine_similarity
```

```python
similarity_scores = cosine_similarity(pt)
```

```python
similarity_scores.shape
```

```python
# Return a list of similar books based on cosine similarity
def recommend(book_name):
    index = np.where(pt.index==book_name)[0][0]
    similar_items = sorted(list(enumerate(similarity_scores[index])),key=lambda x:x[1],reverse=True)[1:5]

    data = []
    for i in similar_items:
        item = []
        temp_df = books[books['Book-Title'] == pt.index[i[0]]]
        item.extend(list(temp_df.drop_duplicates('Book-Title')['Book-Title'].values))
        item.extend(list(temp_df.drop_duplicates('Book-Title')['Book-Author'].values))
        item.extend(list(temp_df.drop_duplicates('Book-Title')['Image-URL-L'].values))

        data.append(item)

    return data
```

```python
recommend('Year of Wonders')
```
